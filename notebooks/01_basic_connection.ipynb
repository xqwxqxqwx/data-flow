```python
import clickhouse_connect

client = clickhouse_connect.get_client(
    host="clickhouse",
    port=8123,
    username="analytics",
    password="analytics",
    database="analytics",
)

client.command("SELECT 1")
```

```python
client.command(
    """
    CREATE TABLE IF NOT EXISTS notebook_demo
    (
      id UInt64,
      value String,
      created_at DateTime DEFAULT now()
    )
    ENGINE = MergeTree
    ORDER BY id
    """
)

client.command("TRUNCATE TABLE notebook_demo")
client.insert("notebook_demo", [(1, "hello"), (2, "data-flow")], column_names=["id", "value"])
```

```python
rows = client.query("SELECT id, value, created_at FROM notebook_demo ORDER BY id").result_rows
rows
```
